3.	Проанализируй количество пустых значений для каждой из групп заявок

```python
# удали из выборки записи той группы, чья доля меньше 5% от общего количества записей
df_1 = df[df['application_type'] != 'JOINT']
```

```python
# удали поля, у которых доля пустых значений на всей выборке превышает 20%
null_ratio = df_1.isna().mean()
high_nulls = null_ratio[null_ratio > 0.2].index
df_clean = df_1.drop(columns = high_nulls)
```

```python
# максимальная доля пустых значений
null_ratio.sort_values()
```

4.	Проанализируй количество дубликатов в оставшихся полях

```python
# Укажи, как часто встречается уникальное значение в поле для ‘grade’, ‘home_ownership’, ‘verification_status’, ‘pymnt_plan’.
df_clean['grade'].value_counts()
```

```python
# Проанализируй оставшиеся столбцы. Удали поля, у которых количество уникальных значений меньше 10
# и распределение значений сильно смещено (доля одного значения от общего числа составляет >80%).

cols = []

for col in df_clean.columns:
  n_unique = df_clean[col].nunique()
  if n_unique < 10:
    top_freq = df_clean[col].value_counts(normalize=True).max()
    if top_freq > 0.8:
      cols.append(col)

print(cols)
df_new = df_clean.drop(columns=cols)
```
