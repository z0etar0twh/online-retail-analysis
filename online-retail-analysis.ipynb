# Занятие 3

```python
country_rev = df_clean.groupby('Country')['Revenue'].sum() #делаем вывод по доминацию UK на рынке
k = country_rev['United Kingdom'] / country_rev.sum()
print(k * 100)
```

**1. Доминация UK на рынке**

UK делает 85% выручки компании -> она сильно централизована.

Риск: при возникновении трудностей на британском рынке вся компания потерпит трудности. "Все яйца в одной корзине" - спад спроса, нечем компенсировать финансово и т.д.

Возможности: расшириться, усилить позиции на глобальном рынке. 15% - огромный потенциал для роста.

**2. Сезонность**

Данные показывают, что пиковая выручка приходится на ноябрь-декабрь - предрождественский сезон.

Риск: не справиться с ажиотажем, потерять клиентов, ошибиться с закупочными позициями и потерять в выручке.

Решение: организовать закупки заранее, изучив рынок трендов и тенденции прошлых лет, нанять дополнительный персонал на время пикового сезона. В *не*пиковые сезоны уплотнить скидки (возможно введение новой системы лояльности, у них высокая эффективность) также заняться маркетингом выгодной распродажи остатков (например, открытие секции прошлого сезона с большими скидками - снизим процент выброшенного). В целом, уплотнить маркетинг, предполагаю недостаток в нем.

**3. Выручка**

Данные показывают, что выручка практически не растет. Выручку есть возможность поднять различными способами - например, сделав упор на расширение в пределах 15% рынка вне Великобритании.

Потенциальные причины: конкуренция, недостаток практик для лояльных клиентов, неизменность неудачных выборок позиций, отсутствие привлекательных предложений и т.д.

Требуется более подробный разбор медианной динамики числа заказов, их состава и чеков.

```python
orders_by_month = df_clean.groupby('YearMonth')['Invoice'].nunique()
orders_by_month
```

```python
revenue_by_month = df_clean.groupby('YearMonth')['Revenue'].sum()
avg_check = revenue_by_month/orders_by_month
avg_check.head()
```

```python
monthly = pd.DataFrame({
    'orders': orders_by_month, #количество заказов
    'revenue': revenue_by_month, #выручка
    'avg_check': avg_check #средний чек
})
monthly
```

```python
quantity_by_month = df_clean.groupby('YearMonth')['Quantity'].sum() #количество айтемов
```

```python
units_per_order = quantity_by_month / orders_by_month #среднее число заказов
price_per_unit = revenue_by_month / quantity_by_month #средняя цена айтема
```

```python
monthly['units_per_order'] = units_per_order
monthly['price_per_unit'] = price_per_unit
monthly
```

```python
monthly['avg_check'].plot(title='Средний чек по месяцам', figsize=(7,4))
```

**Вывод по анализу выручки**

Наблюдение: кол-во заказов оставалось практически неизменным на протяжении периода наблюдения. Результаты анализа среднего чека показали, что в период с конца 2009 по конец 2010 возросли как средняя стоимость одной единицы товара, так и среднее кол-во приобретенного товара в одном чеке. В период конец 2010 - конец 2011 оба показателя вошли в плато.

Диагноз: рост выручки за 2010 год был обусловлен не увеличением количества покупателей и заказов, а увеличением кол-ва товаров в чеке и их стоимости. Рост наблюдался на протяжении 2010 года, однако в 2011 вышел в плато.

Решение: применить стратегии по привлечению новых клиентов. Например, таргетированно рекламироваться на новый диапазон клиентов или расшириться в государствах вне UK (+- 85% области магазина), удовлетворяя их локальные потребности -> увеличить доверие ко всей сети магазинов.

# Занятие 5. SQL

```python
retail = df_clean.copy()
retail['YearMonth'] = retail['YearMonth'].astype(str) # SQLite не понимает тип данных Period
```

```python
import sqlite3 # SQLite - движок для работы с SQL внутри питона

conn = sqlite3.connect(':memory:') # коннект. == кладем в оперативную память, а не по определенному адресу
retail.to_sql('retail', conn, index=False, if_exists='replace') # пандас переносит таблицу в формат SQL. conn - адрес, по которому он ее несет

def q(query):
  return pd.read_sql(query, conn)
```

Подзапросы (subqueries)

```python
# покажи заказы, где выручка строки выше средней по всем строкам
q("""
SELECT Invoice, Description, Revenue
FROM retail
WHERE Revenue > (SELECT AVG(Revenue) FROM retail)
ORDER BY Revenue DESC
LIMIT 10
""")
```

```python
# покажи заказы, где выручка строки выше средней по всем строкам - чистая версия
q("""
SELECT Invoice, Description, Revenue
FROM retail
WHERE (Revenue > (SELECT AVG(Revenue) FROM retail WHERE Description NOT IN ('Manual', 'AMAZON FEE', 'Adjust bad debt', 'POSTAGE', 'DOTCOM POSTAGE'))
AND Description NOT IN ('Manual', 'AMAZON FEE', 'Adjust bad debt', 'POSTAGE', 'DOTCOM POSTAGE'))
ORDER BY Revenue DESC
LIMIT 10
""")
```

```python
q("""
SELECT Invoice, Country, SUM(Revenue) AS order_total
FROM retail
GROUP BY Invoice, Country
""")
```

```python
# найти средний чек по стране — то есть среднюю выручку на один заказ, а не на строку
q("""
SELECT Country, AVG(order_total) AS avg_check
FROM (
  SELECT Invoice, Country, SUM(Revenue) AS order_total
  FROM retail
  GROUP BY Invoice, Country
) AS orders
GROUP BY Country
ORDER BY avg_check DESC
LIMIT 10
""")
```

# Занятие 8. CASE

```python
q("""
SELECT Invoice, Description, Revenue,
  CASE
    WHEN Revenue > 50 THEN 'wholesale'
    ELSE 'retail'
  END AS segment
FROM retail
LIMIT 10
""")
```

```python
# что больше приносит - опт или розница?
q("""
SELECT CASE WHEN Revenue > 50 THEN 'wholesale' ELSE 'retail' END AS segment, SUM(Revenue) AS total_revenue
FROM retail
GROUP BY CASE WHEN Revenue > 50 THEN 'wholesale' ELSE 'retail' END
ORDER BY total_revenue DESC
LIMIT 10
""")
```

```python
q("""
SELECT CASE WHEN Revenue < 20 THEN 'small' WHEN Revenue < 100 THEN 'medium' ELSE 'large' END AS order_size,
  COUNT(*) AS n_lines, SUM(Revenue) AS total_revenue
FROM retail
GROUP BY order_size
ORDER BY total_revenue DESC
""")
```

```python
# в одном запросе, одной строкой результата, сколько выручки дали отдельно возвраты (Revenue < 0) и отдельно продажи (Revenue > 0)
q("""
SELECT SUM(CASE WHEN Revenue > 0 THEN Revenue ELSE 0 END) AS sales,
  SUM(CASE WHEN Revenue < 0 THEN Revenue ELSE 0 END) AS returned
FROM retail
""")
```

```python
# средний чек (выручка на заказ) у «мелких» заказов против «крупных»
q("""
SELECT CASE WHEN order_total < 300 THEN 'small_order' ELSE 'big_order' END AS order_type,
  COUNT(*) AS n_orders, AVG(order_total) AS avg_check
FROM (SELECT Invoice, SUM(Revenue) AS order_total
      FROM retail
      GROUP BY Invoice) AS orders
GROUP BY order_type
ORDER BY avg_check DESC
""")
```
