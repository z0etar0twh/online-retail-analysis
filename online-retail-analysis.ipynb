```python
import pandas as pd

df = pd.read_csv("online_retail_II.csv")
```

```python
df[df['Price'] < 0]
```

```python
df_clean = df[(df['Price'] > 0) & (df['Quantity'] > 0)].copy()
```

```python
df_clean.shape
```

```python
print('min Quantity:', df_clean['Quantity'].min())
print('min Price:', df_clean['Price'].min())
```

```python
df_clean['Revenue'] = df_clean['Price'] * df_clean['Quantity']
```

```python
df_clean['InvoiceDate'] = pd.to_datetime(df_clean['InvoiceDate'])
```

# Занятие 2. Groupby

```python
pd.set_option('display.float_format', '{:,.2f}'.format)
```

```python
df_clean.groupby('Country')['Revenue'].sum().sort_values(ascending=False).head(10)
```

```python
df_clean.groupby('Description')['Revenue'].sum().sort_values(ascending=False).head(10)
```

```python
df_clean['YearMonth'] = df_clean['InvoiceDate'].dt.to_period('M')
```

```python
df_clean[['InvoiceDate', 'YearMonth']].head()
```

```python
monthly = df_clean.groupby('YearMonth')['Revenue'].sum().sort_index()
monthly
```

```python
import matplotlib.pyplot as plt

monthly.plot(kind='line', figsize=(7,3), title='Выручка по месяцам') #можно kind='bar' или style='o'
plt.ylabel('Выручка')
plt.show()
```

# Занятие 3

```python
country_rev = df_clean.groupby('Country')['Revenue'].sum() #делаем вывод по доминацию UK на рынке
k = country_rev['United Kingdom'] / country_rev.sum()
print(k * 100)
```

**1. Доминация UK на рынке**

UK делает 85% выручки компании -> она сильно централизована.

Риск: при возникновении трудностей на британском рынке вся компания потерпит трудности. "Все яйца в одной корзине" - спад спроса, нечем компенсировать финансово и т.д.

Возможности: расшириться, усилить позиции на глобальном рынке. 15% - огромный потенциал для роста.

**2. Сезонность**

Данные показывают, что пиковая выручка приходится на ноябрь-декабрь - предрождественский сезон.

Риск: не справиться с ажиотажем, потерять клиентов, ошибиться с закупочными позициями и потерять в выручке.

Решение: организовать закупки заранее, изучив рынок трендов и тенденции прошлых лет, нанять дополнительный персонал на время пикового сезона. В *не*пиковые сезоны уплотнить скидки (возможно введение новой системы лояльности, у них высокая эффективность) также заняться маркетингом выгодной распродажи остатков (например, открытие секции прошлого сезона с большими скидками - снизим процент выброшенного). В целом, уплотнить маркетинг, предполагаю недостаток в нем.

**3. Выручка**

Данные показывают, что выручка практически не растет. Выручку есть возможность поднять различными способами - например, сделав упор на расширение в пределах 15% рынка вне Великобритании.

Потенциальные причины: конкуренция, недостаток практик для лояльных клиентов, неизменность неудачных выборок позиций, отсутствие привлекательных предложений и т.д.

Требуется более подробный разбор медианной динамики числа заказов, их состава и чеков.
