```python
country_rev = df_clean.groupby('Country')['Revenue'].sum() #делаем вывод по доминацию UK на рынке
k = country_rev['United Kingdom'] / country_rev.sum()
print(k * 100)
```

**1. Доминация UK на рынке**

UK делает 85% выручки компании -> она сильно централизована.

Риск: при возникновении трудностей на британском рынке вся компания потерпит трудности. "Все яйца в одной корзине" - спад спроса, нечем компенсировать финансово и т.д.

Возможности: расшириться, усилить позиции на глобальном рынке. 15% - огромный потенциал для роста.

**2. Сезонность**

Данные показывают, что пиковая выручка приходится на ноябрь-декабрь - предрождественский сезон.

Риск: не справиться с ажиотажем, потерять клиентов, ошибиться с закупочными позициями и потерять в выручке.

Решение: организовать закупки заранее, изучив рынок трендов и тенденции прошлых лет, нанять дополнительный персонал на время пикового сезона. В *не*пиковые сезоны уплотнить скидки (возможно введение новой системы лояльности, у них высокая эффективность) также заняться маркетингом выгодной распродажи остатков (например, открытие секции прошлого сезона с большими скидками - снизим процент выброшенного). В целом, уплотнить маркетинг, предполагаю недостаток в нем.

**3. Выручка**

Данные показывают, что выручка практически не растет. Выручку есть возможность поднять различными способами - например, сделав упор на расширение в пределах 15% рынка вне Великобритании.

Потенциальные причины: конкуренция, недостаток практик для лояльных клиентов, неизменность неудачных выборок позиций, отсутствие привлекательных предложений и т.д.

Требуется более подробный разбор медианной динамики числа заказов, их состава и чеков.

# Занятие 9. Оконные функции

```python
# "сгруппировать по выручке по стране и показать какой % от общего ревенью она занимает" + оконные функции
q("""
SELECT Country,
    SUM(Revenue) AS country_rev,
    SUM(SUM(Revenue)) OVER () AS total_rev,
    ROUND(SUM(Revenue) * 100.0 / SUM(SUM(Revenue)) OVER (), 2) AS pct
FROM retail
GROUP BY Country
ORDER BY country_rev DESC
LIMIT 10
""")
```

```python
# каждый заказ + средний чек по его стране рядом
q("""
SELECT Invoice, Country, order_total,
    ROUND(AVG(order_total) OVER (PARTITION BY Country), 2) AS avg_in_country,
    ROUND(order_total - AVG(order_total) OVER (PARTITION BY Country), 2) AS diff_from_avg
FROM (SELECT Invoice, Country, SUM(Revenue) AS order_total
      FROM retail
      GROUP BY Invoice, Country) AS orders
ORDER BY Country, order_total DESC
LIMIT 20
""")
```

```python
q("""
SELECT *
FROM (SELECT Invoice, Country, SUM(Revenue) AS order_total,
        ROW_NUMBER() OVER (PARTITION BY Country ORDER BY SUM(Revenue) DESC) AS rn
    FROM retail
    GROUP BY Invoice, Country
) AS ranked
WHERE rn < 4
ORDER BY Country, rn
LIMIT 12
""")
```

# Занятие 10

```python
# помесячная выручка с нарастающим итогом

q("""
SELECT YearMonth, SUM(Revenue) AS monthly_rev,
    SUM(SUM(Revenue)) OVER (ORDER BY YearMonth) AS running_total
FROM retail
GROUP BY YearMonth
ORDER BY YearMonth
LIMIT 5
""")
```

```python
# month-over-month: выручка месяца, выручка прошлого месяца рядом, и процент роста

q("""
SELECT YearMonth, monthly_rev,
    LAG(monthly_rev) OVER (ORDER BY YearMonth) AS prev_month,
    ROUND( (monthly_rev - LAG(monthly_rev) OVER (ORDER BY YearMonth)) * 100.0 / LAG(monthly_rev) OVER (ORDER BY YearMonth), 1) AS growth_pct
FROM (SELECT YearMonth, SUM(Revenue) AS monthly_rev
    FROM retail
    GROUP BY YearMonth
) AS monthly
ORDER BY YearMonth
""")
```

```python
# Вопрос 3: продуктовая команда хочет разбить все строки продаж на три корзины по величине Revenue в строке:
# маленькие (<20), средние (от 20 до 100), крупные (>100). Для каждой корзины покажи число строк и суммарную выручку.

q("""
SELECT
    (CASE WHEN Revenue < 20 THEN 'small'
    WHEN Revenue < 100 THEN 'medium'
    ELSE 'big' END) AS size, COUNT(*) AS n_cnt, SUM(Revenue) AS rev
FROM retail
GROUP BY size
""")
```

```python
# Вопрос 4: настоящий средний чек считается по заказам, а не по строкам. Посчитай средний чек (среднюю сумму одного заказа Invoice) по стране
# и выведи топ-10 стран с самым высоким средним чеком. И отдельно ответь словами: почему это нельзя сделать одним простым GROUP BY Country?

q("""
SELECT Country, AVG(total_revenue) AS avg_rev
FROM (SELECT Invoice, Country, SUM(Revenue) AS total_revenue
    FROM retail
    WHERE Description NOT IN ('POSTAGE', 'Manual', 'AMAZON FEE', 'DOTCOM POSTAGE', 'Adjust bad debt')
    GROUP BY Invoice, Country
    ) AS checks
GROUP BY Country
ORDER BY avg_rev DESC
LIMIT 10
""")  # Это нельзя сделать одним простым GROUP BY так как в таком случае среднее считалось бы построчно, а нужно по заказам.
```

```python
# Вопрос 5: для каждой страны найди её топ-3 товара по выручке.
# В результате должны быть видны: страна, товар, его выручка и позиция (1/2/3) внутри страны.

q("""
SELECT Country, StockCode, rev, rn
FROM (SELECT Country, StockCode, SUM(Revenue) AS rev,
    ROW_NUMBER() OVER (PARTITION BY Country ORDER BY SUM(Revenue) DESC) AS rn
    FROM retail
    WHERE Description NOT IN ('POSTAGE', 'Manual', 'AMAZON FEE', 'DOTCOM POSTAGE', 'Adjust bad debt')
    GROUP BY Country, StockCode
    ) AS numbered
WHERE rn < 4
ORDER BY Country, rn
LIMIT 10
""")
```

```python
# bonus: помесячный прирост выручки в % (насколько каждый месяц больше/меньше предыдущего)

q("""
SELECT *
FROM(
  SELECT YearMonth, ym_rev,
    LAG(ym_rev) OVER (ORDER BY YearMonth) AS prev_month,
    ROUND( (ym_rev - LAG(ym_rev) OVER (ORDER BY YearMonth)) * 100.0 / LAG(ym_rev) OVER (ORDER BY YearMonth), 2) AS growth_pct
  FROM (SELECT YearMonth, SUM(Revenue) AS ym_rev
      FROM retail
      GROUP BY YearMonth) AS grouped
  ORDER BY YearMonth )
WHERE growth_pct IS NOT NULL
""")
```
